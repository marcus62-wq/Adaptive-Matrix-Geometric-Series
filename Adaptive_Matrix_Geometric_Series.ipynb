```python

# ============================================================
# PG ADAPTATIVA + GMRES + LU
# MODELO MTL DE SETE CONDUTORES
# GOOGLE COLAB - CODIGO UNICO
#
# Exibe tabelas e Figuras 3 a 12 diretamente no Colab.
# Figura 12: duas superficies 3D e dois graficos 2D.
#
# Referencia LU: verificacao algebrica do modelo MTL.
# Retorno pelo solo: aproximacao, nao Pollaczek integral.
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from pathlib import Path
from time import perf_counter
from scipy.special import ive
from scipy.sparse import lil_matrix, coo_matrix
from scipy.sparse.linalg import splu, gmres, LinearOperator
from IPython.display import display, Markdown

# ============================================================
# 1. PARAMETROS
# ============================================================

OUT = Path("Resultados_PG")
OUT.mkdir(exist_ok=True)

plt.rcParams.update({
    "font.size": 12,
    "font.weight": "bold",
    "axes.labelweight": "bold",
    "axes.titleweight": "bold",
    "figure.facecolor": "white",
    "savefig.facecolor": "white",
    "savefig.dpi": 250,
})

MU0 = 4e-7*np.pi
EPS0 = 8.8541878128e-12

N = 7
F0 = 60.0
I_BASE = 380.0
LENGTH = 4000.0
CELLS = 40

HARMONICS = [1, 3, 5, 7, 9, 11, 13, 15]
GEOMETRIES = ["horizontal", "trefoil"]

SPECTRUM = {
    1: 1.000, 3: 0.070, 5: 0.052, 7: 0.022,
    9: 0.016, 11: 0.006, 13: 0.003, 15: 0.002
}

SOIL_RHO = 100.0
COATING_RHO = 1e10
PIPE_RHO = 2.84e-7
CORE_RHO = 2.42e-8
SHEATH_RHO = 1.02e-7

CORE_RADIUS = 0.0056
SHEATH_R_IN = 0.0133
SHEATH_R_OUT = 0.0151
PIPE_R_IN = 0.1245
PIPE_R_OUT = 0.127
COAT_RADIUS = 0.227

EPSR_INS = 2.634
EPSR_COAT = 3.0

GROUND_R = 1.0
BOND_L = 5e-6
PIPE_ZTERM = 152.0

PG_TOL = 1e-4
PG_KMAX = 128
ALPHAS = [0.25, 0.50, 0.75, 1.00]
GMRES_RTOL = 1e-9

MESH_LEVELS = [20, 40, 80, 160]
MESH_REFERENCE = 320

# ============================================================
# 2. FUNCOES AUXILIARES
# ============================================================

def norm(x):
    return np.linalg.norm(x)

def relative_error(x, ref):
    return 100*norm(x-ref)/max(norm(ref), 1e-30)

def relative_residual(A, x, b):
    return norm(b-A@x)/max(norm(b), 1e-30)

def pipeline(x):
    return x.reshape(-1, N)[:, 6]

def sheath(x, phase=0):
    return x.reshape(-1, N)[:, phase+3]

def heading(text):
    display(Markdown("### " + text))

def show_plot(filename):
    plt.tight_layout()
    plt.savefig(
        OUT/filename, dpi=250,
        bbox_inches="tight", facecolor="white"
    )
    plt.show()
    plt.close()

# ============================================================
# 3. GEOMETRIA
# ============================================================

def layout(kind):
    if kind == "horizontal":
        x = np.array([-0.15, 0.0, 0.15])
        depth = np.ones(3)
    elif kind == "trefoil":
        d = 0.0324
        x = np.array([-d/2, d/2, 0.0])
        depth = np.array([
            1.0, 1.0, 1.0-np.sqrt(3)*d/2
        ])
    else:
        raise ValueError("Geometria desconhecida")

    return np.r_[x, x, 1.0], np.r_[depth, depth, 0.9]

# ============================================================
# 4. IMPEDANCIAS INTERNAS
# ============================================================

def z_internal(rho, radius, omega):
    gamma = np.sqrt(1j*omega*MU0/rho)
    q = gamma*radius
    return (
        rho/(np.pi*radius**2)
        * q/2 * ive(0, q)/ive(1, q)
    )

def z_tube(rho, rin, rout, omega):
    rdc = rho/(np.pi*(rout**2-rin**2))
    q = np.sqrt(1j*omega*MU0/rho)*(rout-rin)
    return rdc*q/np.tanh(q)

# ============================================================
# 5. MATRIZES Z E Y
# ============================================================

def matrices(freq, kind):
    x, depth = layout(kind)
    omega = 2*np.pi*freq

    radii = np.array(
        [CORE_RADIUS]*3 +
        [SHEATH_R_OUT]*3 +
        [PIPE_R_OUT]
    )

    p = np.sqrt(SOIL_RHO/(1j*omega*MU0))
    Z = np.zeros((N, N), dtype=complex)

    for i in range(N):
        for j in range(N):
            if i == j:
                sep = radii[i]
            elif i < 6 and j < 6 and i % 3 == j % 3:
                sep = SHEATH_R_IN
            else:
                sep = max(
                    np.hypot(
                        x[i]-x[j], depth[i]-depth[j]
                    ),
                    radii[i]+radii[j]
                )

            image = np.sqrt(
                (x[i]-x[j])**2
                + (depth[i]+depth[j]+2*p)**2
            )

            Z[i, j] = (
                1j*omega*MU0/(2*np.pi)
                * np.log(image/sep)
            )

    for i in range(3):
        Z[i, i] += z_internal(
            CORE_RHO, CORE_RADIUS, omega
        )

    for i in range(3, 6):
        Z[i, i] += z_tube(
            SHEATH_RHO,
            SHEATH_R_IN,
            SHEATH_R_OUT,
            omega
        )

    Z[6, 6] += z_tube(
        PIPE_RHO, PIPE_R_IN, PIPE_R_OUT, omega
    )

    Y = np.zeros((N, N), dtype=complex)

    ycoax = (
        1j*omega*2*np.pi*EPS0*EPSR_INS
        / np.log(SHEATH_R_IN/CORE_RADIUS)
    )

    for i in range(3):
        j = i+3
        Y[i, i] += ycoax
        Y[j, j] += ycoax
        Y[i, j] -= ycoax
        Y[j, i] -= ycoax

    ycoat = (
        2*np.pi
        * (1/COATING_RHO + 1j*omega*EPS0*EPSR_COAT)
        / np.log(COAT_RADIUS/PIPE_R_OUT)
    )

    Y[6, 6] += ycoat

    for j in range(3, 6):
        Y[j, j] += 1e-10

    return Z, Y

# ============================================================
# 6. MONTAGEM DO SISTEMA MTL
# ============================================================

def build(freq, kind, cells=CELLS):
    Z, Y = matrices(freq, kind)

    dx = LENGTH/cells
    Ys = np.linalg.inv(Z*dx)
    Ysh = Y*dx

    size = (cells+1)*N
    A = lil_matrix((size, size), dtype=complex)

    def stamp(k, l, B):
        for i in range(N):
            for j in range(N):
                if B[i, j] != 0:
                    A[k*N+i, l*N+j] += B[i, j]

    for k in range(cells):
        stamp(k, k, Ys+Ysh/2)
        stamp(k+1, k+1, Ys+Ysh/2)
        stamp(k, k+1, -Ys)
        stamp(k+1, k, -Ys)

    ybond = 1/(
        GROUND_R + 1j*2*np.pi*freq*BOND_L
    )

    for k in [0, cells]:
        for i in range(3, 6):
            A[k*N+i, k*N+i] += ybond

        A[k*N+6, k*N+6] += 1/PIPE_ZTERM

        for i in range(3):
            A[k*N+i, k*N+i] += 1e-5

    b = np.zeros(size, dtype=complex)
    h = round(freq/F0)
    amp = I_BASE*SPECTRUM[h]

    angles = np.array([
        0, -2*np.pi*h/3, 2*np.pi*h/3
    ])

    b[:3] = amp*np.exp(1j*angles)

    return A.tocsc(), b

# ============================================================
# 7. PRECONDICIONADOR
# ============================================================

def block_preconditioner(A):
    C = A.tocoo()
    keep = (C.row % N) == (C.col % N)

    M = coo_matrix(
        (
            C.data[keep],
            (C.row[keep], C.col[keep])
        ),
        shape=A.shape
    ).tocsc()

    return splu(M).solve

# ============================================================
# 8. PG ADAPTATIVA
# ============================================================

def pg_adaptive(A, b, invM, alpha):
    x = np.zeros_like(b)
    history = [relative_residual(A, x, b)]

    best_x = x.copy()
    best_eta = history[0]
    best_k = 0

    target_met = False
    executed = 0
    stagnation = 0

    for k in range(1, PG_KMAX+1):
        r = b-A@x
        step = alpha*invM(r)

        if not np.all(np.isfinite(step)):
            break

        x = x+step
        eta = relative_residual(A, x, b)

        if not np.isfinite(eta):
            break

        history.append(eta)
        executed = k

        if eta < best_eta:
            best_eta = eta
            best_x = x.copy()
            best_k = k

        if eta <= PG_TOL:
            target_met = True
            break

        if eta > history[-2]*1.01:
            stagnation += 1
        else:
            stagnation = 0

        if eta > 1e6 or stagnation >= 12:
            break

    return {
        "x": best_x,
        "alpha": alpha,
        "best_k": best_k,
        "executed": executed,
        "residual": best_eta,
        "history": history,
        "target_met": target_met
    }

# ============================================================
# 9. GMRES
# ============================================================

def gmres_solver(A, b, invM, x0):
    M = LinearOperator(
        A.shape, matvec=invM, dtype=complex
    )

    history = []

    def callback(value):
        history.append(float(value))

    t0 = perf_counter()

    x, info = gmres(
        A, b,
        M=M,
        x0=x0,
        rtol=GMRES_RTOL,
        atol=0,
        restart=40,
        maxiter=20,
        callback=callback,
        callback_type="pr_norm"
    )

    return {
        "x": x,
        "info": info,
        "iterations": len(history),
        "history": history,
        "time": perf_counter()-t0
    }

# ============================================================
# 10. SOLUCAO DE CADA CASO
# ============================================================

def solve_case(kind, h):
    A, b = build(F0*h, kind)

    t0 = perf_counter()
    ref = splu(A).solve(b)
    t_lu = perf_counter()-t0

    t0 = perf_counter()
    invM = block_preconditioner(A)
    t_pre = perf_counter()-t0

    trials = []
    t0 = perf_counter()

    for alpha in ALPHAS:
        trials.append(
            pg_adaptive(A, b, invM, alpha)
        )

    t_search = perf_counter()-t0

    successful = [
        t for t in trials if t["target_met"]
    ]

    if successful:
        chosen = min(
            successful,
            key=lambda t: (t["executed"], t["residual"])
        )
    else:
        chosen = min(
            trials, key=lambda t: t["residual"]
        )

    pure = gmres_solver(
        A, b, invM, np.zeros_like(b)
    )

    corrected = gmres_solver(
        A, b, invM, chosen["x"]
    )

    return {
        "kind": kind,
        "h": h,
        "A": A,
        "b": b,
        "ref": ref,
        "pg": chosen["x"],
        "proposed": corrected["x"],
        "pure": pure["x"],
        "chosen": chosen,
        "trials": trials,
        "pure_info": pure["info"],
        "proposed_info": corrected["info"],
        "pure_iterations": pure["iterations"],
        "proposed_iterations": corrected["iterations"],
        "pure_history": pure["history"],
        "proposed_history": corrected["history"],
        "t_lu": t_lu,
        "t_pre": t_pre,
        "t_search": t_search,
        "t_pure": pure["time"],
        "t_corrected": corrected["time"]
    }

# ============================================================
# 11. METRICAS
# ============================================================

def metrics(r):
    A, b = r["A"], r["b"]
    ref = r["ref"]
    pg = r["pg"]
    prop = r["proposed"]
    pref = pipeline(ref)

    return {
        "Geometry": r["kind"],
        "Harmonic": r["h"],
        "Frequency_Hz": r["h"]*F0,
        "Alpha": r["chosen"]["alpha"],
        "PG_updates": r["chosen"]["best_k"],
        "PG_executed": r["chosen"]["executed"],
        "PG_target_met": r["chosen"]["target_met"],

        "Peak_reference_V": np.max(np.abs(pref)),
        "Peak_PG_V": np.max(np.abs(pipeline(pg))),
        "Peak_proposed_V": np.max(np.abs(pipeline(prop))),

        "PG_global_error_pct": relative_error(pg, ref),
        "PG_pipeline_error_pct": relative_error(
            pipeline(pg), pref
        ),
        "Proposed_global_error_pct": relative_error(
            prop, ref
        ),
        "Proposed_pipeline_error_pct": relative_error(
            pipeline(prop), pref
        ),

        "PG_residual": relative_residual(A, pg, b),
        "Proposed_residual": relative_residual(
            A, prop, b
        ),

        "GMRES_iterations": r["pure_iterations"],
        "PG_GMRES_iterations": r["proposed_iterations"],

        "LU_time_ms": 1000*r["t_lu"],
        "GMRES_time_ms": 1000*(
            r["t_pre"]+r["t_pure"]
        ),
        "PG_GMRES_time_ms": 1000*(
            r["t_pre"]+r["t_search"]+r["t_corrected"]
        ),

        "GMRES_info": r["pure_info"],
        "Proposed_info": r["proposed_info"]
    }

# ============================================================
# 12. FIGURA 3 - TENSAO LONGITUDINAL
# ============================================================

def figure3(results):
    heading("Figura 3 - Third-harmonic pipeline voltage")
    z = np.linspace(0, LENGTH, CELLS+1)

    plt.figure(figsize=(11, 6))

    for kind, color in [
        ("horizontal", "tab:blue"),
        ("trefoil", "tab:orange")
    ]:
        r = results[(kind, 3)]

        for key, style in [
            ("ref", "-"),
            ("pg", "--"),
            ("proposed", ":")
        ]:
            plt.plot(
                z, np.abs(pipeline(r[key])),
                linestyle=style,
                color=color,
                linewidth=2,
                label=f"{kind} {key}"
            )

    plt.xlabel("Position along pipeline (m)")
    plt.ylabel("Voltage magnitude (V)")
    plt.title("Third-harmonic pipeline voltage")
    plt.grid(alpha=0.3)
    plt.legend(ncol=2)
    show_plot("Figura_3.png")

# ============================================================
# 13. FIGURA 4 - RESPOSTA HARMONICA
# ============================================================

def figure4(df):
    heading("Figura 4 - Harmonic response")
    plt.figure(figsize=(11, 6))

    for kind, color in [
        ("horizontal", "tab:blue"),
        ("trefoil", "tab:orange")
    ]:
        d = df[df.Geometry == kind]

        plt.plot(
            d.Harmonic, d.Peak_reference_V,
            "-o", color=color,
            label=f"{kind} Reference"
        )
        plt.plot(
            d.Harmonic, d.Peak_proposed_V,
            "--x", color=color,
            label=f"{kind} Proposed"
        )

    plt.xlabel("Harmonic order")
    plt.ylabel("Peak pipeline voltage (V)")
    plt.title("Harmonic response")
    plt.grid(alpha=0.3)
    plt.legend()
    show_plot("Figura_4.png")

# ============================================================
# 14. FIGURA 5 - CONVERGENCIA DA PG
# ============================================================

def figure5(results):
    heading("Figura 5 - Adaptive PG convergence")
    plt.figure(figsize=(11, 6))

    for h in [1, 3, 9, 15]:
        r = results[("horizontal", h)]
        hist = r["chosen"]["history"]

        plt.semilogy(
            np.arange(len(hist)),
            np.maximum(hist, 1e-17),
            linewidth=2,
            label=f"h = {h}"
        )

    plt.axhline(
        PG_TOL, color="black",
        linestyle=":", linewidth=2,
        label="Tolerance"
    )

    plt.xlabel("Executed PG updates")
    plt.ylabel("True relative residual")
    plt.title("Adaptive PG convergence - horizontal")
    plt.grid(alpha=0.3)
    plt.legend()
    show_plot("Figura_5.png")

# ============================================================
# 15. FIGURA 6 - ATUALIZACOES DA PG
# ============================================================

def figure6(df):
    heading("Figura 6 - Adaptive PG updates")
    plt.figure(figsize=(11, 6))

    for kind, marker in [
        ("horizontal", "o"),
        ("trefoil", "s")
    ]:
        d = df[df.Geometry == kind]
        plt.plot(
            d.Harmonic, d.PG_executed,
            marker=marker, linewidth=2,
            label=kind
        )

    plt.axhline(
        PG_KMAX, color="black",
        linestyle=":", label="Maximum updates"
    )

    plt.xlabel("Harmonic order")
    plt.ylabel("Executed PG updates")
    plt.title("Adaptive PG iteration budget")
    plt.grid(alpha=0.3)
    plt.legend()
    show_plot("Figura_6.png")

# ============================================================
# 16. FIGURAS 7, 8 E 9
# ============================================================

def error_plot(df, pg_col, prop_col,
               title, ylabel, filename):
    plt.figure(figsize=(11, 6))

    for kind in GEOMETRIES:
        d = df[df.Geometry == kind]

        plt.semilogy(
            d.Harmonic,
            np.maximum(d[pg_col], 1e-17),
            "--o", label=f"{kind} PG"
        )
        plt.semilogy(
            d.Harmonic,
            np.maximum(d[prop_col], 1e-17),
            "-s", label=f"{kind} Proposed"
        )

    plt.xlabel("Harmonic order")
    plt.ylabel(ylabel)
    plt.title(title)
    plt.grid(alpha=0.3)
    plt.legend()
    show_plot(filename)

def figures7_8_9(df):
    heading("Figura 7 - Global numerical error")
    error_plot(
        df,
        "PG_global_error_pct",
        "Proposed_global_error_pct",
        "Global numerical error",
        "Relative error (%)",
        "Figura_7.png"
    )

    heading("Figura 8 - Pipeline-specific error")
    error_plot(
        df,
        "PG_pipeline_error_pct",
        "Proposed_pipeline_error_pct",
        "Pipeline-specific numerical error",
        "Relative error (%)",
        "Figura_8.png"
    )

    heading("Figura 9 - Algebraic residual")
    error_plot(
        df,
        "PG_residual",
        "Proposed_residual",
        "Algebraic residual",
        "True relative residual",
        "Figura_9.png"
    )

# ============================================================
# 17. FIGURA 10 - CUSTO COMPUTACIONAL
# ============================================================

def figure10(df):
    heading("Figura 10 - Computational cost")
    plt.figure(figsize=(11, 6))

    for kind in GEOMETRIES:
        d = df[df.Geometry == kind]

        for col, style, label in [
            ("LU_time_ms", "-o", "LU"),
            ("GMRES_time_ms", "--s", "GMRES"),
            ("PG_GMRES_time_ms", ":^", "PG + GMRES")
        ]:
            plt.plot(
                d.Harmonic, d[col],
                style, linewidth=2,
                label=f"{kind} {label}"
            )

    plt.xlabel("Harmonic order")
    plt.ylabel("Execution time (ms)")
    plt.title("Computational cost")
    plt.grid(alpha=0.3)
    plt.legend(ncol=2)
    show_plot("Figura_10.png")

# ============================================================
# 18. FIGURA 11 - RECONSTRUCAO TEMPORAL
# ============================================================

def waveform(results, method, kind="horizontal"):
    t = np.linspace(0, 1/F0, 1000)
    v = np.zeros_like(t)

    for h in HARMONICS:
        phasor = pipeline(
            results[(kind, h)][method]
        )[-1]

        v += np.real(
            phasor*np.exp(1j*2*np.pi*F0*h*t)
        )

    return t, v

def figure11(results):
    heading("Figura 11 - Time-domain reconstruction")
    plt.figure(figsize=(11, 6))

    for method, style in [
        ("ref", "-"),
        ("pg", "--"),
        ("proposed", ":")
    ]:
        t, v = waveform(results, method)

        plt.plot(
            t*1000, v, style,
            linewidth=2,
            label=method.capitalize()
        )

    plt.xlabel("Time (ms)")
    plt.ylabel("Pipeline end voltage (V)")
    plt.title("Time-domain reconstruction")
    plt.grid(alpha=0.3)
    plt.legend()
    show_plot("Figura_11.png")

# ============================================================
# 19. REFINAMENTO ESPACIAL
# ============================================================

def mesh_convergence(kind="horizontal", h=3):
    Aref, bref = build(
        F0*h, kind, cells=MESH_REFERENCE
    )
    xref = splu(Aref).solve(bref)

    zref = np.linspace(
        0, LENGTH, MESH_REFERENCE+1
    )
    vref = pipeline(xref)

    rows = []

    for cells in MESH_LEVELS:
        A, b = build(F0*h, kind, cells=cells)
        x = splu(A).solve(b)

        z = np.linspace(0, LENGTH, cells+1)
        v = pipeline(x)

        vref_interp = (
            np.interp(z, zref, vref.real)
            + 1j*np.interp(z, zref, vref.imag)
        )

        rows.append({
            "Geometry": kind,
            "Harmonic": h,
            "Cells": cells,
            "Step_m": LENGTH/cells,
            "Pipeline_error_pct": relative_error(
                v, vref_interp
            ),
            "Peak_voltage_V": np.max(np.abs(v))
        })

    return pd.DataFrame(rows)

# ============================================================
# 20. FIGURA 12 - QUATRO PAINEIS, DOIS EM 3D
# ============================================================

def figure12(results, mesh_df):
    heading(
        "Figura 12 - Three-dimensional numerical assessment"
    )

    z_km = np.linspace(
        0.0, LENGTH/1000.0, CELLS+1
    )
    hvals = np.asarray(HARMONICS, dtype=float)
    Z, H = np.meshgrid(z_km, hvals)

    reference_map = np.array([
        np.abs(pipeline(results[("horizontal", h)]["ref"]))
        for h in HARMONICS
    ])

    error_map = np.array([
        np.abs(
            pipeline(results[("horizontal", h)]["pg"])
            - pipeline(results[("horizontal", h)]["ref"])
        )
        for h in HARMONICS
    ])

    fig = plt.figure(
        figsize=(17, 12),
        facecolor="white"
    )

    # PAINEL (a) - TENSAO DE REFERENCIA 3D

    ax1 = fig.add_subplot(2, 2, 1, projection="3d")

    surf1 = ax1.plot_surface(
        Z, H, reference_map,
        cmap="viridis",
        linewidth=0,
        edgecolor="none",
        antialiased=True,
        rstride=1,
        cstride=1
    )

    ax1.set_xlabel("Pipeline position (km)", labelpad=8)
    ax1.set_ylabel("Harmonic order", labelpad=8)
    ax1.set_zlabel("Voltage magnitude (V)", labelpad=8)
    ax1.set_yticks(HARMONICS)
    ax1.set_title("(a) Reference pipeline voltage", pad=16)
    ax1.view_init(elev=26, azim=-60)

    fig.colorbar(
        surf1, ax=ax1,
        shrink=0.60, pad=0.14,
        label="Voltage magnitude (V)"
    )

    # PAINEL (b) - ERRO DA PG 3D

    ax2 = fig.add_subplot(2, 2, 2, projection="3d")

    surf2 = ax2.plot_surface(
        Z, H, error_map,
        cmap="magma",
        linewidth=0,
        edgecolor="none",
        antialiased=True,
        rstride=1,
        cstride=1
    )

    ax2.set_xlabel("Pipeline position (km)", labelpad=8)
    ax2.set_ylabel("Harmonic order", labelpad=8)
    ax2.set_zlabel("Absolute error (V)", labelpad=8)
    ax2.set_yticks(HARMONICS)
    ax2.set_title("(b) Adaptive PG spatial error", pad=16)
    ax2.view_init(elev=26, azim=-60)

    fig.colorbar(
        surf2, ax=ax2,
        shrink=0.60, pad=0.14,
        label="Absolute voltage error (V)"
    )

    # PAINEL (c) - RECONSTRUCAO TEMPORAL

    ax3 = fig.add_subplot(2, 2, 3)

    for method, label, style in [
        ("ref", "Reference", "-"),
        ("pg", "Adaptive PG", "--"),
        ("proposed", "Proposed", ":")
    ]:
        t, v = waveform(
            results, method, "horizontal"
        )

        ax3.plot(
            t*1000.0, v,
            linestyle=style,
            linewidth=2.2,
            label=label
        )

    ax3.set_xlabel("Time (ms)")
    ax3.set_ylabel("Pipeline end voltage (V)")
    ax3.set_title("(c) Time-domain reconstruction")
    ax3.grid(alpha=0.30)
    ax3.legend()

    # PAINEL (d) - CONVERGENCIA ESPACIAL

    ax4 = fig.add_subplot(2, 2, 4)

    d = mesh_df.sort_values("Step_m")
    steps = d["Step_m"].to_numpy(dtype=float)
    errors = d["Pipeline_error_pct"].to_numpy(dtype=float)

    valid = (
        np.isfinite(steps)
        & np.isfinite(errors)
        & (steps > 0)
        & (errors > 0)
    )

    ax4.loglog(
        steps[valid],
        errors[valid],
        "-o",
        linewidth=2.2,
        markersize=7,
        label="Numerical error"
    )

    if valid.sum() >= 3:
        slope, intercept = np.polyfit(
            np.log(steps[valid]),
            np.log(errors[valid]),
            1
        )

        xfit = np.sort(steps[valid])
        yfit = np.exp(intercept)*xfit**slope

        ax4.loglog(
            xfit, yfit, "--",
            linewidth=1.8,
            label=f"Observed slope = {slope:.2f}"
        )

        print(
            "Observed mesh-refinement slope:",
            f"{slope:.4f}"
        )
        print(
            "A inclinacao e empirica e nao constitui "
            "prova de ordem assintotica."
        )

    ax4.set_xlabel("Spatial step (m)")
    ax4.set_ylabel("Pipeline relative error (%)")
    ax4.set_title("(d) Spatial mesh refinement")
    ax4.grid(alpha=0.30, which="both")
    ax4.legend()

    fig.suptitle(
        "Numerical assessment of harmonic-induced "
        "pipeline voltages",
        fontsize=18,
        fontweight="bold",
        y=0.98
    )

    fig.subplots_adjust(
        left=0.05,
        right=0.94,
        bottom=0.07,
        top=0.90,
        wspace=0.30,
        hspace=0.33
    )

    fig.savefig(
        OUT/"Figura_12_3D.png",
        dpi=300,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.show()
    plt.close(fig)

# ============================================================
# 21. TABELAS EXIBIDAS NO COLAB
# ============================================================

def display_tables(df, mesh_df, results):
    pd.set_option("display.max_columns", None)
    pd.set_option("display.width", 200)

    heading("Tabela 1 - Resultados numericos completos")
    display(df)

    summary = df.groupby("Geometry").agg(
        PG_max_error_pct=("PG_global_error_pct", "max"),
        PG_pipeline_max_error_pct=(
            "PG_pipeline_error_pct", "max"
        ),
        Proposed_max_error_pct=(
            "Proposed_global_error_pct", "max"
        ),
        PG_max_residual=("PG_residual", "max"),
        Proposed_max_residual=(
            "Proposed_residual", "max"
        ),
        Mean_LU_ms=("LU_time_ms", "mean"),
        Mean_GMRES_ms=("GMRES_time_ms", "mean"),
        Mean_PG_GMRES_ms=("PG_GMRES_time_ms", "mean")
    )

    heading("Tabela 2 - Resumo estatistico")
    display(summary)

    convergence = df[[
        "Geometry", "Harmonic", "Alpha",
        "PG_updates", "PG_executed",
        "PG_target_met", "PG_residual"
    ]].copy()

    heading("Tabela 3 - Convergencia da PG")
    display(convergence)

    trial_rows = []

    for (kind, h), r in results.items():
        for trial in r["trials"]:
            trial_rows.append({
                "Geometry": kind,
                "Harmonic": h,
                "Alpha": trial["alpha"],
                "Best_updates": trial["best_k"],
                "Executed_updates": trial["executed"],
                "Residual": trial["residual"],
                "Tolerance_met": trial["target_met"]
            })

    trials_df = pd.DataFrame(trial_rows)

    heading("Tabela 4 - Busca de alpha")
    display(trials_df)

    heading("Tabela 5 - Refinamento espacial")
    display(mesh_df)

    return summary, convergence, trials_df

# ============================================================
# 22. EXECUCAO PRINCIPAL
# ============================================================

def main():
    heading("PG adaptativa + GMRES - Analise completa")

    results = {}
    rows = []

    for kind in GEOMETRIES:
        for h in HARMONICS:
            r = solve_case(kind, h)
            results[(kind, h)] = r

            row = metrics(r)
            rows.append(row)

            print(
                f"{kind:10s} | "
                f"h={h:2d} | "
                f"alpha={row['Alpha']:.2f} | "
                f"PG updates={row['PG_updates']:3d} | "
                f"PG residual={row['PG_residual']:.3e} | "
                f"Proposed residual="
                f"{row['Proposed_residual']:.3e}"
            )

    df = pd.DataFrame(rows)

    if (
        (df.GMRES_info != 0).any()
        or (df.Proposed_info != 0).any()
    ):
        print(
            "AVISO: alguma execucao GMRES nao "
            "atingiu o criterio interno."
        )

    print("\nExecutando refinamento espacial...")

    mesh_df = mesh_convergence(
        kind="horizontal", h=3
    )

    summary, convergence, trials_df = display_tables(
        df, mesh_df, results
    )

    # FIGURAS NA ORDEM DO MANUSCRITO

    figure3(results)
    figure4(df)
    figure5(results)
    figure6(df)
    figures7_8_9(df)
    figure10(df)
    figure11(results)
    figure12(results, mesh_df)

    # PLANILHA OPCIONAL - AS TABELAS JA FORAM EXIBIDAS

    try:
        with pd.ExcelWriter(
            OUT/"Resultados_Completos.xlsx",
            engine="openpyxl"
        ) as writer:
            df.to_excel(
                writer, sheet_name="Results", index=False
            )
            summary.to_excel(
                writer, sheet_name="Summary"
            )
            convergence.to_excel(
                writer, sheet_name="PG_Convergence",
                index=False
            )
            trials_df.to_excel(
                writer, sheet_name="Alpha_Search",
                index=False
            )
            mesh_df.to_excel(
                writer, sheet_name="Mesh_Refinement",
                index=False
            )
    except ImportError:
        print(
            "openpyxl indisponivel. "
            "As tabelas continuam visiveis."
        )

    print("\n========================================")
    print("RESUMO FINAL")
    print("========================================")

    print("Casos analisados:", len(df))
    print(
        "PG atingiu tolerancia:",
        int(df.PG_target_met.sum())
    )
    print(
        "Erro global maximo PG (%):",
        df.PG_global_error_pct.max()
    )
    print(
        "Erro maximo tubulacao PG (%):",
        df.PG_pipeline_error_pct.max()
    )
    print(
        "Erro global maximo Proposed (%):",
        df.Proposed_global_error_pct.max()
    )
    print(
        "Media iteracoes GMRES:",
        df.GMRES_iterations.mean()
    )
    print(
        "Media iteracoes PG + GMRES:",
        df.PG_GMRES_iterations.mean()
    )

    print("Total de figuras: 10")
    print("Pasta de resultados:", OUT)

    return df, results, mesh_df

# ============================================================
# 23. EXECUTAR TUDO
# ============================================================

DF, RESULTS, MESH_DF = main()
```
